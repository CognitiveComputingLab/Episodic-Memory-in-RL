```python
import os
from pathlib import Path

from xmaze_jetcobblestone import generate_dataset, save_dataset_to_disk, XMazeEnv, OptimalXMazePolicy, compute_dataset_info

min_instructions = 1
max_instructions = 2
possible_instructions = 2
min_corridor_len = 6
max_corridor_len = 7
episodes = 64
instruction_every = 1
clip = None
encoding = 'onehot_repeat'


env = XMazeEnv(
    turns_range=range(min_instructions,max_instructions),
    colour_count=possible_instructions,
    init_corridor_range=range(min_corridor_len,max_corridor_len),
    instruction_every=instruction_every,
    clip=clip,
    obs_mode=encoding,
)

dataset = generate_dataset(env, episodes, 1000, OptimalXMazePolicy(env))
ds_info = compute_dataset_info([episode["observations"] for episode in dataset])

location = Path("data", "XMaze", env.getDatasetName(
    min_instructions=min_instructions,
    max_instructions=max_instructions,
    possible_instructions=possible_instructions,
    min_corridor_len=min_corridor_len,
    max_corridor_len=max_corridor_len,
    episodes=episodes,
    instruction_every=instruction_every,
    clip=None,
    encoding=encoding
))
print(location)
```

```python
save_dataset_to_disk(dataset, ds_info, location)
```
